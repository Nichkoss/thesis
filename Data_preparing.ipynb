<h3>Context</h3> 

<p>Kidney cancer remains a significant challenge in modern medicine despite notable progress in diagnostic and treatment methods. The issue of predicting the risk of disease progression is particularly relevant, as even after radical treatment, some patients experience recurrences or metastases. These circumstances complicate the selection of optimal treatment strategies and significantly impact patients' quality of life and survival rates.</p>

<h3>Research Object</h3>

<p>The process of predicting the risk of kidney cancer progression based on clinical individual characteristics of patients and data from patients treated at the Lviv Oncology Center between 2007 and 2018.</p>

<h3>Research Subject</h3>

<p>The use of machine learning methods to build a model that predicts the risk of disease progression, taking into account individual patient characteristics (age, gender, tumor process stage, histological and immunohistochemical study results).</p>

<h3>Research Goal</h3>

<p>The goal of the research is to develop a model that predicts the risk of kidney cancer progression using machine learning methods to improve treatment strategies and enhance therapy efficiency.</p>

<h3>Relevance</h3> 
<p>The relevance of the study lies in the fact that, despite modern medical advancements and the availability of various treatment methods, kidney cancer (KC) remains a significant problem. Most cases are diagnosed at a localized stage, where the tumor has not yet spread beyond the kidney. However, even after radical treatment, disease progression occurs in a significant proportion of patients during follow-up, manifesting as local recurrences, distant metastases, or multifocal kidney cancer.</p> <p>Particular attention is required for patients with locally advanced KC, including cases with tumor extension to perirenal fat, the adrenal gland, or the presence of tumor thrombus in the renal or inferior vena cava. The presence of isolated metastases in regional lymph nodes also significantly increases the risk of adverse outcomes and complicates treatment strategy selection, as these patients face higher risks of complications and disease progression.</p> <p>Additionally, patients with distant metastases at initial diagnosis often present challenges in disease management. Progression is typically observed during systemic therapy evaluation (targeted therapy, immunotherapy), complicating the choice of further treatment strategies. Literature data indicate that 30-40% of radically treated patients experience progression, with the risk of recurrence or metastasis reaching up to 50% in some cases, emphasizing the necessity of risk prediction based on individual patient characteristics.</p> <p>Thus, developing a model that predicts the risk of disease progression based on individual patient characteristics—such as age, gender, tumor stage, and results of pathological and immunohistochemical studies—is an essential task. This approach will help clinicians make more informed decisions regarding treatment strategies and disease control, provide patients with accurate prognostic information, and enhance therapy efficiency while reducing mortality rates from this condition.</p>
